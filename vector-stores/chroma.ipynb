```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
from langchain_core.documents import Document

# Create LangChain documents for IPL players

doc1 = Document(
    page_content="Virat Kohli is an Indian cricketer and former captain of the Indian national team. He is widely regarded as one of the best batsmen in the world.",
    metadata={"player": "Virat Kohli", "team": "Royal Challengers Bangalore", "role": "Batsman"}
)

doc2 = Document(
    page_content="MS Dhoni is a former Indian cricketer and captain of the Indian national team. He is known for his calm demeanor and exceptional leadership skills.",
    metadata={"player": "MS Dhoni", "team": "Chennai Super Kings", "role": "Wicketkeeper-Batsman"}
)

doc3 = Document(
    page_content="Rashid Khan is an Afghan cricketer who plays for the Afghanistan national team. He is a leg-spin bowler and is known for his exceptional bowling skills.",
    metadata={"player": "Rashid Khan", "team": "Sunrisers Hyderabad", "role": "Bowler"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is an Indian cricketer who plays for the Indian national team. He is a fast bowler and is known for his unique bowling action and ability to bowl yorkers.",
    metadata={"player": "Jasprit Bumrah", "team": "Mumbai Indians", "role": "Bowler"}
)

doc5 = Document(
    page_content="Rohit Sharma is an Indian cricketer and the current captain of the Indian national team. He is known for his aggressive batting style and ability to score big centuries.",
    metadata={"player": "Rohit Sharma", "team": "Mumbai Indians", "role": "Batsman"}
)
```

```python
# Create a list of all documents
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
# Add documents to Chroma vector store

vector_store = Chroma(
    embedding_function=HuggingFaceEmbeddings(),
    persist_directory="my_chroma_db",
    collection_name='sample'
)
```

```python
# Generate unique IDs for each document and add them to the vector store

vector_store.add_documents(docs)
```

```python
# Retrieve documents from the vector store with embeddings, documents, and metadata 

vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
# search documents in the vector store based on a query and return the top k most similar documents

vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
```

### Here Similarity Score tell the distance between vectors if distance is low then similarity between two vector is more and if distance is more then similarity between two vector is low

```python
# search with similarity score and return the top k most similar documents

vector_store.similarity_search_with_score(
    query='Who among these are a bowler?',
    k=2
)
```

```python
# meta-deta filtering to retrieve documents based on specific metadata criteria

vector_store.similarity_search_with_score(
    query='',
    k=2,
    filter={"team": "Mumbai Indians"}
)
```

```python
# update documents in the vector store by providing a list of updated documents with the same IDs as the existing documents

updated_doc1 = Document(
    page_content="Virat Kohli is an Indian cricketer and former captain of the Indian national team. He is widely regarded as one of the best batsmen in the world. He has also played for the Royal Challengers Bangalore in the Indian Premier League.",
    metadata={"player": "Virat Kohli", "team": "Royal Challengers Bangalore", "role": "Batsman"}
)

vector_store.update_document(document_id='356477b0-6b6f-435c-86fc-d25377773c58', document=updated_doc1)
```

```python
# view the updated document in the vector store

vector_store.get(include=['embeddings', 'documents', 'metadatas'])
```

```python
# delete a document from the vector store by providing its unique ID

vector_store.delete(ids=['356477b0-6b6f-435c-86fc-d25377773c58'])
```
